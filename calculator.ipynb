```python
def add(a, b):
  return a + b
def subst(a, b):
  return a - b
def multip(a, b):
  return a * b
def div(a, b):
  if b == 0:
    print("cannot divide by zero , choose another number: ")
  b = float(input())
  return a / b
def expo(a, b):
  return a ** b
calculation = 0
while True:
 print("python calculator")
 print("1 : add")
 print("2 : subst")
 print("3 : multip")
 print("4 : div")
 print("5 : quit")
 print("6: exponantiation")

 choice = input("choose a option: ")
 if choice == "5" :
   print("Goodbye !")
   print(f"you have done", calculation, "calculations")
   break

 num1 = float(input("Enter your firt number: "))
 num2 = float(input("Enter your second number"))

 if choice == "1":
    print(add(num1, num2))
    calculation += 1
 elif choice == "2":
    print(subst(num1, num2))
    calculation += 1
 elif choice == "3":
     print(multip(num1, num2))
     calculation += 1
 elif choice == "4":
     print(div(num1, num2))
     calculation += 1
 elif choice == "6":
     print(expo(a=num1, b=num2))
     calculation += 1

 else:
      print("invalid choice")




```
