```python
import requests
import random
url = "https://countriesnow.space/api/v0.1/countries/capital"

response = requests.get(url)

data = response.json()


countries_list = data["data"]

quiz = {}

for item in countries_list:
    name = item["name"]
    capital = item["capital"]

    if capital:
        quiz[name] = capital


countries = list(quiz.keys())
capitals = list(quiz.values())

difficulty = input("Choose you difficulty level easy , medium , hard")
if difficulty == 'easy':
  d = 5
elif difficulty == "medium":
  d = 10
elif difficulty == " hard":
  d = 20

score = 0
question_number = 0
wrong_answers = 0
countries_missed = []

while question_number < d:
  country = random.choice(countries)
  countries.remove(country)
  capital = quiz[country]
  print("press q if you want to quit ")
  answer = input(f"What is the capital of {country} ?")
  if answer.lower() == 'q':
    break
  elif answer.lower() == capital.lower():
    score += 1
    question_number = question_number + 1
    print("Correct ")
  else:
    question_number = question_number + 1
    wrong_answers += 1
    countries_missed.append(country)
    print("Wrong !")
    print("The answer was", capital)

print("The final score is",score)
print("Countries missed:", countries_missed)
```
